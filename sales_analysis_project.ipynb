# Sales Analysis Project
# Objective: Analyze customer spending and revenue patterns

This project analyzes customer transaction data to identify revenue patterns, customer behavior, and business opportunities to improve sales performance.

Business Questions:
1. Which day generates the highest revenue?
2. Which time period is most profitable?
3. Which customer segment contributes more?
4. What are the tipping behaviors of customers?

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/tips.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

**No missing values found**

## **Data Analysis**

Total Revenue

```python
df.total_bill.sum()
```

Revenue by Day

```python
df.groupby('day')['total_bill'].sum()
```

Revenue by Time

```python
df.groupby('time')['total_bill'].sum()
```

Average Bill

```python
df['total_bill'].mean()
```

Gender Spending

```python
df.groupby('sex')['total_bill'].mean()
```

Tip Percentage

```python
df['tip _percentage']=(df['tip']/df['total_bill'])*100
df.groupby('sex')['tip _percentage'].mean()
```

**Bar Chart**

```python
df.groupby('day')['total_bill'].sum().plot(kind='bar')
plt.title('Revenue Day')
plt.show()
```

**Pie Chart**

```python
df['sex'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.show()
```

**Histogram**

```python
df['total_bill'].hist()
plt.show()
```

# **Conclusion:**
- Weekend dinners are the peak business time.
- Female customers show higher tipping behavior in percentage.
- Businesses should focus on weekend promotions to maximize revenue.
